```python
# CELL 2: Import Libraries
# =============================================================================
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from textblob import TextBlob
from vaderSentiment.vaderSentiment import SentimentIntensityAnalyzer
# Topic Modeling
from gensim import corpora
from gensim.models import LdaModel, CoherenceModel
from gensim.utils import simple_preprocess
# Text preprocessing
import re
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
from collections import Counter
# Visualization
from wordcloud import WordCloud
import pyLDAvis
import pyLDAvis.gensim_models as gensimvis
import warnings
warnings.filterwarnings('ignore')
print("✅ All libraries imported successfully!")
```

```python
# CELL 3: Load the Data
# =============================================================================
# For Google Colab - upload your file
from google.colab import files
print("Please upload your fakenews.csv file:")
uploaded = files.upload()
# Load the CSV
df = pd.read_csv('fakenews.csv')
# Clean up - the CSV has extra empty columns
df = df[['text', 'label']].dropna()
print(f"\n📊 Dataset Shape: {df.shape}")
print(f"\n📋 Label Distribution:")
print(df['label'].value_counts())
print(f"\n🔍 Sample Data:")
df.head()
```

```python
# CELL 4: Text Preprocessing Function
# =============================================================================
def preprocess_text(text):
    """Clean and preprocess text for analysis."""
    if not isinstance(text, str):
        return ""

    # Convert to lowercase
    text = text.lower()

    # Remove URLs
    text = re.sub(r'http\S+|www\S+|https\S+', '', text)

    # Remove special characters and numbers
    text = re.sub(r'[^a-zA-Z\s]', '', text)

    # Remove extra whitespace
    text = ' '.join(text.split())

    return text
# Apply preprocessing
df['clean_text'] = df['text'].apply(preprocess_text)
print("✅ Text preprocessing complete!")
print(f"\nSample cleaned text:\n{df['clean_text'].iloc[0][:500]}...")
```

```python
# PART 1: SENTIMENT ANALYSIS
# =============================================================================
# =============================================================================
# CELL 5: Compute Sentiment Scores
# =============================================================================
print("="*60)
print("PART 1: SENTIMENT ANALYSIS")
print("="*60)
# Initialize VADER sentiment analyzer (better for social media/news text)
vader = SentimentIntensityAnalyzer()
def get_vader_sentiment(text):
    """Get VADER sentiment scores."""
    if not isinstance(text, str) or len(text) == 0:
        return {'neg': 0, 'neu': 0, 'pos': 0, 'compound': 0}
    return vader.polarity_scores(text)
def get_textblob_sentiment(text):
    """Get TextBlob sentiment (polarity and subjectivity)."""
    if not isinstance(text, str) or len(text) == 0:
        return 0, 0
    blob = TextBlob(text)
    return blob.sentiment.polarity, blob.sentiment.subjectivity
# Compute VADER sentiment
print("Computing VADER sentiment scores...")
vader_scores = df['text'].apply(get_vader_sentiment)
df['vader_compound'] = vader_scores.apply(lambda x: x['compound'])
df['vader_positive'] = vader_scores.apply(lambda x: x['pos'])
df['vader_negative'] = vader_scores.apply(lambda x: x['neg'])
df['vader_neutral'] = vader_scores.apply(lambda x: x['neu'])
# Compute TextBlob sentiment
print("Computing TextBlob sentiment scores...")
textblob_scores = df['text'].apply(get_textblob_sentiment)
df['textblob_polarity'] = textblob_scores.apply(lambda x: x[0])
df['textblob_subjectivity'] = textblob_scores.apply(lambda x: x[1])
# Categorize sentiment
def categorize_sentiment(compound):
    if compound >= 0.05:
        return 'Positive'
    elif compound <= -0.05:
        return 'Negative'
    else:
        return 'Neutral'
df['sentiment_category'] = df['vader_compound'].apply(categorize_sentiment)
print("✅ Sentiment analysis complete!")
```

```python
# CELL 6: Compare Sentiment: Spam vs Non-Spam
# =============================================================================
# Create labels for clarity
df['label_name'] = df['label'].map({0: 'Real News', 1: 'Fake News (Spam)'})
# Summary statistics by label
print("\n" + "="*60)
print("SENTIMENT COMPARISON: FAKE NEWS vs REAL NEWS")
print("="*60)
sentiment_summary = df.groupby('label_name').agg({
    'vader_compound': ['mean', 'std', 'median'],
    'vader_positive': 'mean',
    'vader_negative': 'mean',
    'textblob_polarity': ['mean', 'std'],
    'textblob_subjectivity': 'mean'
}).round(4)
print(sentiment_summary)
# Statistical test
from scipy import stats
real_news_sentiment = df[df['label'] == 0]['vader_compound']
fake_news_sentiment = df[df['label'] == 1]['vader_compound']
t_stat, p_value = stats.ttest_ind(real_news_sentiment, fake_news_sentiment)
print(f"\n📊 Statistical Test (Independent t-test):")
print(f"   t-statistic: {t_stat:.4f}")
print(f"   p-value: {p_value:.6f}")
if p_value < 0.05:
    print("   ✅ Significant difference in sentiment between groups!")
else:
    print("   ❌ No significant difference in sentiment between groups.")
```

```python
# CELL 7: Visualize Sentiment Results
# =============================================================================
fig, axes = plt.subplots(2, 2, figsize=(14, 10))
# 1. Box plot of compound scores
ax1 = axes[0, 0]
df.boxplot(column='vader_compound', by='label_name', ax=ax1)
ax1.set_title('VADER Compound Sentiment by News Type')
ax1.set_xlabel('News Type')
ax1.set_ylabel('Compound Sentiment Score')
plt.suptitle('')
# 2. Distribution of sentiment
ax2 = axes[0, 1]
for label in df['label_name'].unique():
    subset = df[df['label_name'] == label]['vader_compound']
    ax2.hist(subset, bins=50, alpha=0.5, label=label, density=True)
ax2.set_title('Distribution of Sentiment Scores')
ax2.set_xlabel('VADER Compound Score')
ax2.set_ylabel('Density')
ax2.legend()
# 3. Sentiment category counts
ax3 = axes[1, 0]
sentiment_counts = df.groupby(['label_name', 'sentiment_category']).size().unstack(fill_value=0)
sentiment_counts.plot(kind='bar', ax=ax3, rot=0)
ax3.set_title('Sentiment Categories by News Type')
ax3.set_xlabel('News Type')
ax3.set_ylabel('Count')
ax3.legend(title='Sentiment')
# 4. Mean sentiment comparison
ax4 = axes[1, 1]
mean_sentiments = df.groupby('label_name')[['vader_positive', 'vader_negative', 'vader_neutral']].mean()
mean_sentiments.plot(kind='bar', ax=ax4, rot=0)
ax4.set_title('Average Sentiment Components by News Type')
ax4.set_xlabel('News Type')
ax4.set_ylabel('Mean Score')
ax4.legend(title='Sentiment Type')
plt.tight_layout()
plt.savefig('sentiment_analysis.png', dpi=150, bbox_inches='tight')
plt.show()
print("\n📈 Sentiment visualization saved as 'sentiment_analysis.png'")
```

```python
# CELL 8: Sentiment Analysis Conclusions
# =============================================================================
print("\n" + "="*60)
print("📋 SENTIMENT ANALYSIS CONCLUSIONS")
print("="*60)
real_mean = df[df['label'] == 0]['vader_compound'].mean()
fake_mean = df[df['label'] == 1]['vader_compound'].mean()
print(f"\n📊 Average Sentiment Scores:")
print(f"   Real News:  {real_mean:.4f}")
print(f"   Fake News:  {fake_mean:.4f}")
if fake_mean > real_mean:
    print(f"\n✅ FINDING: Fake news has HIGHER sentiment ({fake_mean:.4f}) than real news ({real_mean:.4f})")
    print("   This suggests fake news tends to be more emotionally charged/positive.")
elif fake_mean < real_mean:
    print(f"\n✅ FINDING: Fake news has LOWER sentiment ({fake_mean:.4f}) than real news ({real_mean:.4f})")
    print("   This suggests fake news tends to be more negative or neutral.")
else:
    print("\n✅ FINDING: No significant difference in sentiment between fake and real news.")
```

```python
# PART 2: TOPIC MODELING
# =============================================================================
# =============================================================================
# CELL 9: Prepare Data for Topic Modeling
# =============================================================================
print("\n" + "="*60)
print("PART 2: TOPIC MODELING")
print("="*60)
# Preprocessing for topic modeling
stop_words = set(stopwords.words('english'))
lemmatizer = WordNetLemmatizer()
# Add custom stop words
custom_stopwords = {'said', 'would', 'could', 'also', 'one', 'two', 'new',
                    'like', 'get', 'make', 'even', 'say', 'first', 'go',
                    'us', 'year', 'people', 'time', 'know', 'take'}
stop_words = stop_words.union(custom_stopwords)
def preprocess_for_lda(text):
    """Preprocess text for LDA topic modeling."""
    if not isinstance(text, str):
        return []

    # Tokenize
    tokens = simple_preprocess(text, deacc=True, min_len=3)

    # Remove stopwords and lemmatize
    tokens = [lemmatizer.lemmatize(token) for token in tokens
              if token not in stop_words and len(token) > 2]

    return tokens
# Apply preprocessing
print("Preprocessing text for topic modeling...")
df['tokens'] = df['clean_text'].apply(preprocess_for_lda)
# Filter out empty documents
df_lda = df[df['tokens'].apply(len) > 5].copy()
print(f"Documents for topic modeling: {len(df_lda)}")
# Create dictionary and corpus
texts = df_lda['tokens'].tolist()
dictionary = corpora.Dictionary(texts)
# Filter extremes
dictionary.filter_extremes(no_below=20, no_above=0.5)
# Create corpus
corpus = [dictionary.doc2bow(text) for text in texts]
print(f"✅ Dictionary size: {len(dictionary)} unique tokens")
print(f"✅ Corpus size: {len(corpus)} documents")
```

```python
# CELL 10: Find Optimal Number of Topics
# =============================================================================
print("\n🔍 Finding optimal number of topics using coherence scores...")
def compute_coherence_values(dictionary, corpus, texts, limit, start=2, step=2):
    """Compute coherence values for different numbers of topics."""
    coherence_values = []
    model_list = []

    for num_topics in range(start, limit, step):
        model = LdaModel(
            corpus=corpus,
            id2word=dictionary,
            num_topics=num_topics,
            random_state=42,
            passes=10,
            alpha='auto',
            per_word_topics=True
        )
        model_list.append(model)
        coherence_model = CoherenceModel(
            model=model,
            texts=texts,
            dictionary=dictionary,
            coherence='c_v'
        )
        coherence_values.append(coherence_model.get_coherence())
        print(f"   Topics: {num_topics}, Coherence: {coherence_values[-1]:.4f}")

    return model_list, coherence_values
# Test different numbers of topics
limit = 16
start = 3
step = 2
model_list, coherence_values = compute_coherence_values(
    dictionary, corpus, texts, limit, start, step
)
# Plot coherence scores
plt.figure(figsize=(10, 5))
x = range(start, limit, step)
plt.plot(x, coherence_values, 'b-o')
plt.xlabel("Number of Topics")
plt.ylabel("Coherence Score (c_v)")
plt.title("Topic Coherence Scores for Different Numbers of Topics")
plt.xticks(x)
plt.grid(True, alpha=0.3)
plt.savefig('topic_coherence.png', dpi=150, bbox_inches='tight')
plt.show()
# Find optimal
optimal_idx = np.argmax(coherence_values)
optimal_topics = start + optimal_idx * step
print(f"\n✅ OPTIMAL NUMBER OF TOPICS: {optimal_topics}")
print(f"   Coherence score: {coherence_values[optimal_idx]:.4f}")
```

```python
# CELL 11: Train Final LDA Model
# =============================================================================
print(f"\nTraining final LDA model with {optimal_topics} topics...")
# Use the best model or retrain
lda_model = model_list[optimal_idx]
# Print topics
print("\n" + "="*60)
print(f"TOP {optimal_topics} TOPICS DISCOVERED")
print("="*60)
for idx, topic in lda_model.print_topics(num_topics=optimal_topics, num_words=10):
    print(f"\n📌 Topic {idx + 1}:")
    # Parse and format the topic words
    words = re.findall(r'"([^"]+)"', topic)
    weights = re.findall(r'(0\.\d+)', topic)
    for word, weight in zip(words[:10], weights[:10]):
        print(f"   • {word}: {float(weight):.3f}")
```

```python
print("\n" + "="*60)
print("TOPIC DISTRIBUTION BY NEWS TYPE")
print("="*60)
# Get dominant topic for each document
def get_dominant_topic(bow, lda_model):
    """Get the dominant topic for a document."""
    topic_probs = lda_model.get_document_topics(bow)
    if topic_probs:
        dominant = max(topic_probs, key=lambda x: x[1])
        return dominant[0], dominant[1]
    return None, 0
# Get topic for each document
topics_data = []
for i, (bow, row_idx) in enumerate(zip(corpus, df_lda.index)):
    topic_num, topic_prob = get_dominant_topic(bow, lda_model)
    topics_data.append({
        'dominant_topic': topic_num,
        'topic_probability': topic_prob,
        'label': df_lda.loc[row_idx, 'label'],
        'label_name': df_lda.loc[row_idx, 'label_name']
    })
topics_df = pd.DataFrame(topics_data)
# Topic distribution by label
print("\n📊 Topic Distribution by News Type:\n")
topic_label_crosstab = pd.crosstab(
    topics_df['dominant_topic'],
    topics_df['label_name'],
    normalize='columns'
).round(4) * 100
print(topic_label_crosstab)
print("\n(Values shown as percentages of each news type)")
# Visualize
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
# Topic distribution by news type
ax1 = axes[0]
topic_label_crosstab.plot(kind='bar', ax=ax1, rot=0)
ax1.set_title('Topic Distribution by News Type (%)')
ax1.set_xlabel('Topic Number')
ax1.set_ylabel('Percentage of Documents')
ax1.legend(title='News Type')
# Chi-square test for topic-label relationship
contingency = pd.crosstab(topics_df['dominant_topic'], topics_df['label'])
chi2, p_value, dof, expected = stats.chi2_contingency(contingency)
ax2 = axes[1]
# Heatmap of topic-label relationship
sns.heatmap(contingency, annot=True, fmt='d', cmap='YlOrRd', ax=ax2)
ax2.set_title(f'Topic-Label Contingency Table\n(Chi² = {chi2:.2f}, p = {p_value:.4f})')
ax2.set_xlabel('News Label (0=Real, 1=Fake)')
ax2.set_ylabel('Topic Number')
plt.tight_layout()
plt.savefig('topic_distribution.png', dpi=150, bbox_inches='tight')
plt.show()
print(f"\n📊 Chi-Square Test for Topic-Label Independence:")
print(f"   Chi² statistic: {chi2:.4f}")
print(f"   p-value: {p_value:.6f}")
print(f"   Degrees of freedom: {dof}")
if p_value < 0.05:
    print("   ✅ Topics are SIGNIFICANTLY related to fake/real news classification!")
else:
    print("   ❌ No significant relationship between topics and news classification.")
```
